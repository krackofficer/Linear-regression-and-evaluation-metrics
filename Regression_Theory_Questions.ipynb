# Theoretical Regression Questions

```python
# 1. What does R-squared represent in a regression model?
print('''R-squared represents the proportion of variance in the dependent variable that is predictable from the independent variable(s).''')
```

```python
# 2. What are the assumptions of linear regression?
print('''Assumptions include linearity, independence of errors, homoscedasticity, normality of residuals, and no multicollinearity.''')
```

```python
# 3. What is the difference between R-squared and Adjusted R-squared?
print('''R-squared measures explained variance, while Adjusted R-squared adjusts for the number of predictors to avoid overfitting.''')
```

```python
# 4. Why do we use Mean Squared Error (MSE)?
print('''MSE measures the average squared difference between actual and predicted values, helping to quantify prediction error.''')
```

```python
# 5. What does an Adjusted R-squared value of 0.85 indicate?
print('''An Adjusted R-squared of 0.85 means 85% of the variance is explained, adjusted for the number of predictors used.''')
```

```python
# 6. How do we check for normality of residuals in linear regression?
print('''We can check normality using Q-Q plots, histograms of residuals, or statistical tests like the Shapiro-Wilk test.''')
```

```python
# 7. What is multicollinearity, and how does it impact regression?
print('''Multicollinearity occurs when predictors are highly correlated, which inflates standard errors and weakens statistical power.''')
```

```python
# 8. What is Mean Absolute Error (MAE)?
print('''MAE measures the average magnitude of errors in predictions, without considering direction.''')
```

```python
# 9. What are the benefits of using an ML pipeline?
print('''ML pipelines streamline workflows by automating preprocessing, training, and evaluation, ensuring reproducibility and efficiency.''')
```

```python
# 10. Why is RMSE considered more interpretable than MSE?
print('''RMSE is in the same units as the target variable, making it easier to interpret than squared errors.''')
```

```python
# 11. What is pickling in Python, and how is it useful in ML?
print('''Pickling serializes Python objects, useful for saving ML models for reuse without retraining.''')
```

```python
# 12. What does a high R-squared value mean?
print('''A high R-squared value indicates the model explains a large portion of the variance, but it doesn’t guarantee accuracy.''')
```

```python
# 13. What happens if linear regression assumptions are violated?
print('''Violating assumptions can lead to biased, inconsistent, or inefficient estimates.''')
```

```python
# 14. How can we address multicollinearity in regression?
print('''Multicollinearity can be reduced using PCA, dropping variables, or using regularization techniques like Ridge or Lasso.''')
```

```python
# 15. How can feature selection improve model performance in regression analysis?
print('''Feature selection removes irrelevant features, reducing overfitting and improving model interpretability and performance.''')
```

```python
# 16. How is Adjusted R-squared calculated?
print('''Adjusted R-squared = 1 - [(1 - R²)(n - 1) / (n - k - 1)], where n is sample size and k is number of predictors.''')
```

```python
# 17. Why is MSE sensitive to outliers?
print('''MSE squares errors, so it heavily penalizes larger errors, making it sensitive to outliers.''')
```

```python
# 18. What is the role of homoscedasticity in linear regression?
print('''Homoscedasticity ensures constant variance of residuals, which is essential for valid hypothesis tests and confidence intervals.''')
```

```python
# 19. What is Root Mean Squared Error (RMSE)?
print('''RMSE is the square root of MSE and represents the average error magnitude in the same unit as the dependent variable.''')
```

```python
# 20. Why is pickling considered risky?
print('''Pickled files may execute malicious code if tampered with, posing a security risk.''')
```

```python
# 21. What alternatives exist to pickling for saving ML models?
print('''Alternatives include using `joblib`, saving models in ONNX or PMML format, or exporting to JSON when applicable.''')
```

```python
# 22. What is heteroscedasticity, and why is it a problem?
print('''Heteroscedasticity means residuals have non-constant variance, leading to inefficient estimates and invalid inferences.''')
```

```python
# 23. How can interaction terms enhance a regression model's predictive power?
print('''Interaction terms capture the combined effect of two variables, improving model performance when relationships aren't purely additive.''')
```
